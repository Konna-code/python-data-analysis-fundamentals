## 3. If-Statements with User-Defined Functions

```python
# Defining the price range function.
```

```python
def price_label(row):

  if row['prices'] <= 5:
    return 'Low-range product'
  elif (row['prices'] > 5) and (row['prices'] <= 15):
    return 'Mid-range product'
  elif row['prices'] > 15:
    return 'High range'
  else: return 'Not enough data'
```

```python
# Applying the function
```

```python
df['price_range'] = df.apply(price_label, axis=1)
```

```python
# Checking results with the value_counts functions
```

```python
df['price_range'].value_counts(dropna=False)
```

```python
df['prices'].max()
```

## 4. If-Statements with the loc() Function

```python
# Creating some conditions for the price range using the loc() function.
```

```python
df.loc[df['prices'] > 15, 'price_range_loc'] = 'High-range product'
```

```python
df.loc[(df['prices'] <= 15) & (df['prices'] > 5), 'price_range_loc'] = 'Mid-range product'
```

```python
df.loc[df['prices'] <= 5, 'price_range_loc'] = 'Low-range product'
```

```python
# Checking results
```

```python
df['price_range_loc'].value_counts(dropna=False)
```

```python
# Now running it on the entire data frame
```

```python
df_ords_prods_merge.loc[df_ords_prods_merge['prices'] > 15, 'price_range_loc'] = 'High-range product'
```

## 2. Creating 'busiest days' column that indicates the two busiest and two slowest days.

```python
# Checking the frequency of orders placed during the week days.
```

```python
df_ords_prods_merge['orders_day_of_week'].value_counts(dropna=False)
```

## Day 0 and 1 are the busiest days, while day 3 and 4 are the slowest days. Day 6,2 and 5 are regular days. 

```python
# Executing the code to create the new list.
```

```python
result = []

for value in df_ords_prods_merge["orders_day_of_week"]:
  if value in (0,1):
    result.append("Two busiest days")
  elif value in (3,4):
    result.append("Two slowest days")
  else:
    result.append("Regular days")
```

```python
# Now replacing the values of the previous busiest_day column.
```

```python
df_ords_prods_merge['busiest_day']=result
```

```python
# Renaming the column from 'busiest_day' to 'busiest_days'
```

```python
df_ords_prods_merge.rename(columns={'busiest_day':'busiest_days'},inplace=True)
```

```python
df_ords_prods_merge['busiest_days'].value_counts(dropna=False)
```

## 3. Observations: naturally the busiest and slowest days increased as it now contains the values of two days each. This can be cross checked by adding day 0 and 1 together, as well as 3 and 4 in the frequency check.

## 4. Creating a new column indicating the 'busiest period of the day'

```python
# Checking the frequency of the purchase hours to identify the bussiest periods. 
```

```python
df_ords_prods_merge['hour_of_purchase'].value_counts(dropna=False)
```

```python
# Creating 3 conditions for the hour of purchase range using the loc() function.
```

```python
result = []

for value in df_ords_prods_merge["hour_of_purchase"]:
  if value in (10,11,14,15,13,12,16,9):
    result.append("Most orders,")
  elif value in (17,8,18,19,20,7,21,22):
    result.append("Average orders,")
  elif value in (23,6,0,1,5,2,4,3):
    result.append("Fewest orders.")
  else:
    result.append("N/A")
```

```python
df_ords_prods_merge['busiest_period_of_day']=result
```
